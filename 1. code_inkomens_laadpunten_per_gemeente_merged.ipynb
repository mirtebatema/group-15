```python
import pandas as pd

# Inlezen
laad = pd.read_csv("Aantal laadpunten per gemeente.csv", sep=";")
laad.columns = ["gemeente", "laadpunten"]

ink = pd.read_csv("Inkomens.csv", sep=";", encoding="utf-8-sig")
ink = ink.rename(columns={"Regio's": "gemeente"})

# Getallen staan als tekst met spaties ervoor -> omzetten naar getallen
for kolom in ink.columns[4:]:
    ink[kolom] = pd.to_numeric(ink[kolom].astype(str).str.strip(), errors="coerce")

# Namen gelijk maken: CBS-naam -> naam uit het laadpuntenbestand
naam_mapping = {
    "'s-Gravenhage (gemeente)": "Den Haag",
    "Groningen (gemeente)": "Groningen",
    "Utrecht (gemeente)": "Utrecht",
    "Beek (L.)": "Beek",
    "Stein (L.)": "Stein",
    "Hengelo (O.)": "Hengelo",
    "Laren (NH.)": "Laren",
    "Middelburg (Z.)": "Middelburg",
    "Rijswijk (ZH.)": "Rijswijk",
}
ink["gemeente"] = ink["gemeente"].replace(naam_mapping)

# Regel "Totaal" weghalen
laad = laad[laad["gemeente"] != "Totaal"]

# Samenvoegen
data = laad.merge(ink, on="gemeente", how="inner")

data = data[["gemeente", "laadpunten", "Populatie", "Perioden", "Particuliere huishoudens (x 1 000)", "Inkomen/Gemiddeld gestandaardiseerd inkomen (1 000 euro)"]]
print(data.shape)  # (342, 10)
```
