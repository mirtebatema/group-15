```python
sales = [
    [120, 80, 200, 50, 160],
    [90, 110, 70, 180, 130],
    [250, 200, 90, 300, 150],
    [60, 40, 100, 80, 120]
]

for i in range(len(sales)):
    total = 0
    highest = 0

    for j in range(len(sales[i])):
        if sales[i][j] >= 150:
            total += 1

        if sales[i][j] > highest:
            highest = sales[i][j]

    print(i + 1, total, highest)
```
