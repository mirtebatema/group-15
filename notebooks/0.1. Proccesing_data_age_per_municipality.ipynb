# Age per municipality

This notebook merges the CBS data on young people (aged 0–25) and older people (aged 65+) per municipality and saves the result for the other notebooks.

**Input files (`data/raw/`):**
- `jongeren_per_gemeente_2026.csv`: percentage of residents aged 0–25 per municipality (2026).
- `ouderen_per_gemeente_2026.csv`: percentage of residents aged 65+ per municipality (2026).

**Output (`data/processed/`):**
- `age_per_municipality.csv`: one row per municipality (342), containing the percentages of young and older residents.

```python
import pandas as pd
```

```python
# Load the data
young = pd.read_csv("../data/raw/jongeren_per_gemeente_2026.csv", sep=";", decimal=",", encoding="utf-8-sig")
older = pd.read_csv("../data/raw/ouderen_per_gemeente_2026.csv", sep=";", decimal=",", encoding="utf-8-sig")

# Select the relevant columns and give them English names
young = young[["Gemeente", "Totaal, 0 tot 25"]].copy()
young.columns = ["municipality", "pct_young_0_25"]
older = older[["Gemeente", "Totaal, 65 jaar of ouder"]].copy()
older.columns = ["municipality", "pct_older_65plus"]

# Merge the two datasets
data = young.merge(older, on="municipality", how="inner", validate="one_to_one")
print(data.shape)  # Expected: (342, 3)

# Save
data.to_csv("../data/processed/age_per_municipality.csv", index=False)
```
