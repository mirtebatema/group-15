```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt

# ==========================================
# 1. JOUW DATA INLEZEN
# ==========================================

df = pd.read_csv(
    "/Users/mirtebatema/Desktop/TIL_2026/programming/opdracht/laadpunten_inkomen_per_gemeente.csv",
    sep=None,
    engine="python"
)

display(df.head())

# Alleen relevante kolommen
data = df[
    ["gemeente", "laadpunten_per_1000_huishoudens"]
].copy()

display(data.head())


# ==========================================
# 2. LANDKAART MET GEMEENTEGRENZEN INLEZEN
# ==========================================

gemeenten = gpd.read_file(
    "gemeentegrenzen.geojson"
)

display(gemeenten.head())
print(gemeenten.columns)


# ==========================================
# 3. DATA KOPPELEN AAN DE LANDKAART
# ==========================================

kaart = gemeenten.merge(
    data,
    left_on="gemeentenaam",   # naam van gemeente in GeoJSON
    right_on="gemeente",
    how="left"
)


# ==========================================
# 4. KAART MAKEN
# ==========================================

fig, ax = plt.subplots(figsize=(10, 12))

kaart.plot(
    column="laadpunten_per_1000_huishoudens",
    cmap="Blues",
    linewidth=0.3,
    edgecolor="white",
    legend=True,
    ax=ax,
    missing_kwds={
        "color": "lightgrey",
        "label": "Geen data"
    }
)

ax.set_title(
    "Aantal publieke laadpunten per 1.000 huishoudens",
    fontsize=16
)

ax.axis("off")

plt.tight_layout()
plt.show()
```
